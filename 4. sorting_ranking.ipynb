# Sorting and Ranking (တန်ဖိုးများ စီခြင်းနှင့် အဆင့်သတ်မှတ်ခြင်း)

## 1. Sorting (ကြီးစဉ်ငယ်လိုက် သို့ ငယ်စဉ်ကြီးလိုက်စီခြင်း)

```python
import numpy as np
import pandas as pd
```

## 2. အဆင့်သတ်မှတ်ခြင်း (Ranking)

```python
# Series တစ်ခုကို index များရောထွေး၍ တည်ဆောက်ခြင်း
obj = pd.Series(np.arange(4), index=["d", "a", "b", "c"])
obj
```

```python
obj.sort_index()
```

```python
obj.sort_index(ascending=False)
```

```python
# DataFrame တစ်ခု တည်ဆောက်ခြင်း
frame = pd.DataFrame(np.arange(8).reshape((2, 4)),
                     index=["three", "one"],
                     columns=["d", "a", "b", "c"])
frame
```

```python
frame.sort_index()
```

```python
frame.sort_index(axis="columns", ascending=False)
```

### တန်ဖိုးများအလိုက် စီခြင်း (Sorting by Values)

```python
obj = pd.Series([4, 7, -3, 2])
obj
```

```python
obj.sort_values()
```

```python
obj = pd.Series([4, np.nan, 7, np.nan, -3, 2])
obj
```

```python
obj.sort_values()
```

```python
obj.sort_values(na_position="first")
```

```python
# နမူနာ DataFrame တည်ဆောက်ခြင်း
frame = pd.DataFrame({"b": [4, 7, -3, 2], "a": [0, 1, 0, 1]})
frame
```

```python
frame.sort_values("b")
```

```python
frame.sort_values("a")
```

```python
frame.sort_values(["a", "b"])
```

```python
frame.sort_values(["a", "b"]).reset_index(drop=True)
```

## 2 အဆင့်သတ်မှတ်ခြင်း (Ranking)

```python
# Series တစ်ခု တည်ဆောက်ခြင်း
obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj
```

```python
obj.rank()
```

```python
obj.rank(method="first")
```

```python
obj.rank(ascending=False)
```

```python
# DataFrame အသစ်တစ်ခု တည်ဆောက်ခြင်း
frame = pd.DataFrame({"b": [4.3, 7, -3, 2], 
                      "a": [0, 1, 0, 1],
                      "c": [-2, 5, 8, -2.5]})
frame
```

```python
frame.rank()
```

```python
frame.rank(axis="columns")
```
