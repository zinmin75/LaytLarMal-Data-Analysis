# Essential Functionality (Pandas ရဲ့ အခြေခံလုပ်ဆောင်ချက်များ)

သည် video မှာ Series နဲ့ DataFrame နဲ့ ပတ်သတ်ပြီး data များနှင့် အပြန်အလှန်လုပ်ဆောင်ရာတွင် အသုံးပြုရမယ့် အခြေခံအကျဆုံး နည်းလမ်းများကို လေ့လာသွားမည် ဖြစ်ပါတယ်။ နောက်ဗီဒီယိုတွေမှာတော့ pandas ကိုအသုံးပြုပြီး data analysis နဲ့ data များကိုင်တွယ်ခြင်း (manipulation) ဆိုင်ရာ အကြောင်းအရာတွေကို ပိုပြီး အတွင်းကျကျ လေ့လာသွားမှာ ဖြစ်ပါတယ်။ 
သည် ဗီဒီယိုကတော့ pandas library တစ်ခုလုံးကို အသေးစိတ်မှတ်တမ်း (documentation) တင်ဖို့ ရည်ရွယ်ထားခြင်းမဟုတ်ဘဲ အသုံးအများဆုံး feature များကိုသာ အဓိကထားဖော်ပြပေးမှာ ဖြစ်ပါတယ်။ သိပ်ပြီးအသုံးမများသော အကြောင်းအရာတွေကိုတော့ pandas ရဲ့ online documentation မှာ ထပ်မံဖတ်ရှုနိုင်ပါတယ်။

```python
import pandas as pd
import numpy as np
```

# 1. Reindexing (Index ပြန်လည်သတ်မှတ်ခြင်း)

## 1.1 Series များတွင် Reindex လုပ်ခြင်း

```python
# Series တစ်ခုကို index သတ်မှတ်ပြီး တည်ဆောက်ခြင်း
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=["d", "b", "a", "c"])
obj
```

```python
obj2 = obj.reindex(["a", "b", "c", "d", "e"])
obj2
```

Time series လိုမျိုး အစဉ်လိုက်ရှိတဲ့ data တွေကို reindex လုပ်သည့်အခါ တန်ဖိုးတွေကို ဖြည့်စွက်ခြင်း (interpolation/filling) ပြုလုပ်လိုနိုင်တယ်။ `method` option ကို အသုံးပြုပြီး လုပ်ဆောင်နိုင်ပြီး `ffill` (forward-fill) ကတော့ ပျောက်နေတဲ့ တန်ဖိုးတွေကို ရှေ့မှာရှိတဲ့ တန်ဖိုးကို ယူပြီး ဆက်တိုက်ဖြည့်ပေးသွားမည် ဖြစ်တယ်။

```python
obj3 = pd.Series(["blue", "purple", "yellow"], index=[0, 2, 4])
obj3
```

```python
np.arange(6)
```

```python
# method="ffill" ကိုသုံးပြီး ကွက်လပ်များကို ရှေ့တန်ဖိုးဖြင့် ဖြည့်ခြင်း
obj3.reindex(np.arange(6))
```

```python
obj3.reindex(np.arange(6), method="ffill")
```

## 1.2 DataFrame များတွင် Reindex လုပ်ခြင်း

```python
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=["a", "c", "d"],
                     columns=["Ohio", "Texas", "California"])
frame
```

```python
# Row index များကို reindex လုပ်ခြင်း
frame2 = frame.reindex(index=["a", "b", "c", "d"])
frame2
```

```python
states = ["Texas", "Utah", "California"]
frame.reindex(columns=states)
```

```python
frame
```

```python
frame.reindex(states, axis="columns")
```

```python
frame.reindex(states, axis=1)
```

### reindex function ၏ အဓိက argument များ

| Argument | Description |
| :--- | :--- |
| **labels** | index အဖြစ် အသုံးပြုမည့် sequence အသစ် ဖြစ်သည်။ |
| **index** | index (rows) အသစ်အဖြစ် အသုံးပြုမည့် sequence ဖြစ်သည်။ |
| **columns** | column label အသစ်အဖြစ် အသုံးပြုမည့် sequence ဖြစ်သည်။ |
| **axis** | reindex လုပ်မည့် axis ကို သတ်မှတ်သည်။ "index" (rows) သို့မဟုတ် "columns" ဖြစ်နိုင်သည်။ |
| **method** | အချက်အလက်ဖြည့်စွက်သည့်နည်းလမ်း (Interpolation method) ဖြစ်သည်။ "ffill" (forward) သို့မဟုတ် "bfill" (backward) ဖြစ်နိုင်သည်။ |
| **fill_value** | reindex လုပ်စဉ် data မရှိသောနေရာများတွင် အစားထိုးမည့် တန်ဖိုးဖြစ်သည်။ |

`loc` operatorကို အသုံးပြုပြီးလည်း reindex လုပ်နိုင်တယ်။ သည်နည်းလမ်းက index label အသစ်တွေ အကုန်လုံးက DataFrame အတွင်းထဲ ရှိနေမှသာ အလုပ်လုပ်မှာ ဖြစ်တယ်။

```python
frame
```

```python
frame.loc[["a", "d", "c"], ["California", "Texas"]]
```

```python
frame.loc[["a", "d", "c", "asdfs"], ["California", "Texas"]]
```

## 3.1 Series တွင် Indexing လုပ်ခြင်း

Series indexing (`obj[...]`) က NumPy array indexing နဲ့ ဆင်တူပေမယ့်လည်း integer(location) များအပြင် Series ရဲ့ index label တွေကိုပါ အသုံးပြုနိုင်ခြင်းမှာ ကွာခြားချက်ဖြစ်တယ်။

```python
obj = pd.Series(np.arange(4.), index=["a", "b", "c", "d"])
obj
```

```python
obj["b"]
```

```python
obj[1]
```

```python
# Slicing လုပ်ခြင်း
obj[2:4]
```

```python
obj["b":"d"]
```

### `loc` နှင့် `iloc` အော်ပရေတာများ

Label တွေ သုံးပြီး data indexing လုပ်ရင် အပေါ်ကလိုမျိုး indexing မဟုတ်ပဲ သီးသန့်လုပ်ထားတဲ့ `loc` operator ကို အသုံးပြုဖို့ ပိုပြီးအကြံပြုတယ်။ အဲ့လို operator မျိုး သုံးခြင်းဖြင့် indexing လုပ်ရာမှာ integer တွေကို label အဖြစ် သတ်မှတ်ခြင်း နဲ့ position အဖြစ် သတ်မှတ်ခြင်းတို့မှာ ဖြစ်ပေါ်တတ်တဲ့ ရှုပ်ထွေးမှုတွေကို ရှောင်ရှားနိုင်မှာ ဖြစ်ပါတယ်။

```python
obj
```

```python
obj.loc["b"]
```

```python
obj.loc[["b", "a", "d"]]
```

အကယ်၍ index မှာ integer တွေ ပါဝင်နေပါက `[]` သုံးပြီး indexing လုပ်ခြင်းက label နှင့် position ကြားမှာ ရှုပ်ထွေးနိုင်တယ်။ ထို့ကြောင့် position အခြေပြု indexing အတွက် `iloc` ကို အသုံးပြုသင့်တယ်။

```python
obj1 = pd.Series([1, 2, 3], index=[2, 0, 1])
obj2 = pd.Series([1, 2, 3], index=["a", "b", "c"])
```

```python
obj1
```

```python
obj2
```

```python
obj1.iloc[[0, 1, 2]]
```

```python
obj1.loc[2]
```

> **သတိပြုရန်:** Label များနဲ့ slicing လုပ်ရာမှာ (Python ရဲ့ ပုံမှန် slicing နဲ့မတူဘဲ) အဆုံးသတ် point (endpoint) သည် inclusive ဖြစ်တယ် (ဆိုလိုသည်မှာ အဆုံးသတ် label ပါ ပါဝင်တယ်)။

```python
obj2
```

```python
# Label slicing (endpoint inclusive ဖြစ်သည်)
obj2.loc["b":"c"]
```

### 3.2 DataFrame တွင် Indexing လုပ်ခြင်း

DataFrame မှာ index လုပ်ခြင်းကတော့ column တစ်ခု သို့မဟုတ် တစ်ခုထက်ပိုတဲ့ column တွေကို ရွေးချယ်ပေးတယ်။

```python
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
data
```

```python
data["two"]
```

- slicing ဆိုရင် အဲ့လိုလုပ်တာ column slice မဟုတ်တော့ပဲ row တွေကို slice တယ်။
- Boolean Indexing (filtering)

```python
# Row များကို slicing လုပ်ခြင်း (အဆင်ပြေစေရန် ပြုလုပ်ပေးထားသော syntax ဖြစ်သည်)
data[:2]
```

### `loc` နှင့် `iloc` ကို အသုံးပြု၍ DataFrame တွင် Data ရွေးချယ်ခြင်း

Series လိုပဲ DataFrame မှာလည်း label အခြေပြု indexing အတွက် `loc` နဲ့ integer(position) အခြေပြု indexing အတွက် `iloc` ဟူသော အထူး attribute များ ရှိတယ်။ DataFrame က two-dimensional ဖြစ်သောကြောင့် NumPy လို ပုံစံနဲ့ row ကော column ကောကို label သို့မဟုတ် integer သုံးပြီး indexing လုပ်နိုင်တယ်။

အရင် label ကို အသုံးပြုပြီး row တစ်ခုတည်းကို ရွေးချယ်ပုံကို ကြည့်ရအောင်။

```python
# နမူနာ DataFrame တည်ဆောက်ခြင်း
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
data
```

```python
# Label ကိုသုံး၍ row တစ်ခုတည်းကို ရွေးချယ်ခြင်း
# ရလဒ်သည် column label များပါဝင်သော Series တစ်ခု ဖြစ်လာပါမည်။
data.loc["Colorado"]
```

```python
data.loc[["Colorado", "New York"]]
```

`loc` တွင် row နှင့် column ရွေးချယ်မှုများကို ကော်မာ (comma) ဖြင့် ခွဲခြား၍ တစ်ပြိုင်နက် လုပ်ဆောင်နိုင်ပါသည်။

```python
# Row နှင့် column ကို တွဲဖက်ရွေးချယ်ခြင်း
data.loc["Colorado", "three"]
```

```python
data.loc["Colorado", ["two", "three"]]
```

```python
data.loc[["Colorado", "Utah"], ["two", "three"]]
```

ဆက်လက်၍ `iloc` ကို အသုံးပြု၍ integer များဖြင့် အလားတူ ရွေးချယ်မှုများကို လုပ်ဆောင်ကြည့်ပါမည်။

```python
data
```

```python
# Integer position ဖြင့် row တစ်ခုကို ရွေးချယ်ခြင်း
data.iloc[2]
```

```python
# Integer position စာရင်းဖြင့် row များကို ရွေးချယ်ခြင်း
data.iloc[[2, 1]]
```

```python
# Row နှင့် column position များကို တွဲဖက်ရွေးချယ်ခြင်း
data.iloc[2, [3, 0, 1]]
```

```python
# Row အများအပြားနှင့် column အများအပြားကို position ဖြင့် ရွေးချယ်ခြင်း
data.iloc[[1, 2], [3, 0, 1]]
```

Indexing attribute နှစ်ခုလုံးသည် single label များအပြင် slices များနှင့်လည်း အလုပ်လုပ်ပါသည်။

```python
data
```

```python
# loc ဖြင့် slicing လုပ်ခြင်း
data.loc[:"Utah", "two"]
```

```python
data.iloc[:, :3]
```

```python
data.iloc[:, :3][data['three'] > 5]
```

```python
data.iloc[:, :3][data['three'] > 5].loc["Utah"].loc["two"]
```

> **သတိပြုရန်:** Boolean array များကို `loc` နှင့် တွဲဖက်အသုံးပြုနိုင်သော်လည်း `iloc` နှင့် တွဲဖက်အသုံးပြု၍ မရပါ။

```python
# Boolean array ကို loc ဖြင့် အသုံးပြုခြင်း
data.loc[data.three >= 2]
```

```python
data.iloc[data.three >= 2]
```

# 5. Function Application and Mapping (Function များ အသုံးပြုခြင်းနှင့် Mapping လုပ်ခြင်း)

```python
# random standard normal values များဖြင့် DataFrame တည်ဆောက်ခြင်း
frame = pd.DataFrame(np.random.standard_normal((4, 3)),
                     columns=list("bde"),
                     index=["Utah", "Ohio", "Texas", "Oregon"])
frame
```

```python
# element-wise အဖြစ် absolute values (အနှုတ်လက္ခဏာဖြုတ်ခြင်း) များ ရယူခြင်း - Vectorization
np.abs(frame)
```

```python
frame
```

```python
# Series တစ်ခု၏ maximum နှင့် minimum ခြားနားချက်ကို ရှာသည့် function f1 -> range
def f1(x):
    return x.max() - x.min()
```

```python
frame.apply(f1)
```

```python
frame.apply(f1, axis="columns")
```

```python
# min နှင့် max တန်ဖိုးနှစ်ခုလုံးကို Series အဖြစ် ပြန်ပေးမည့် function f2
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])
```

```python
frame.apply(f2, axis="columns")
```

```python
# ဒသမ ၂ နေရာယူရန် format သတ်မှတ်သည့် function
def my_format(x):
    return f"{x:.2f}"
```

```python
frame.map(my_format)
```
