### 1.1 Series 
Column တစ်ခုသာ ရှိတဲ့ Data Structure မျိုး

. 
NumPy အနေနဲ့ ကြည့်မယ်ဆိုရင်တော့ 1-D Array (Vector

```python
obj.array
```

```python
obj.index
```

`.array` attribute ကို သုံးပြီး ကြည့်ရင် pandas ရဲ့ array တွေက NumPy array တွေကို အမှီပြုထားတာ တွေ့ရတယ်။

တစ်ခါတစ်လေ data တစ်ခုချင်းစီကို default index မဟုတ်ပဲ သပ်သပ် label လေးတွေ ပေးချင်တယ်။ (NumPy မှာ မရ)

"Utah" ကတော့ states ထဲမှာ မပါဝင်နေလို့ result အသစ်မှာ မပါလာဘူး။


အဲ့လို missing data တွေကို ဒီ playlist မှာ “missing,” “NA,” or “null” ဆိုပြီးတော့ ခေါ်ဝေါ်သွားမယ်။

```python
pd.isna(obj4)
```

```python
pd.notna(obj4)
```

Series မှာ ဒီဟာတွေကို instance method အနေနဲ့လည်း ပါရှိတယ်။

### 1.2 DataFrame
- Row နဲ့ Column တွေ ပါတဲ့ Table (Spreadsheet) ပုံစံ Data Structure မျိုး
- NumPy အနေနဲ့ ကြည့်မယ်ဆိုရင်တော့ 2-D Array (Matrix) ပေါ့
- index တွေတူတဲ့ Series(Columns) တွေ စုထားတဲ့ Table လို့ ပြောလို့ရတယ်
- DataFrame မှာ ကျတော့ row ကော column ကော index တွေ ရှိတ်။ ယ်။


Note: DataFrame က တည်ဆောက်ပုံအားဖြင့် 2-D ဆိုပေမယ့် hierarchical indexing သဘောတရားကို သုံးပြီး higher-dimensional table ပုံစံတွေ ဆောက်လို့ ရတယ်။

DataFrame တွေ ပြုလုပ်လို့ ရတဲ့ နည်းလမ်း အမျိုးမျိုး ရှိတယ်။ အသုံးအများဆုံးကတော့ အရွယ်အစားတူညီတဲ့ list သို့မဟုတ် NumPy array တွေနဲ့ ဖွဲ့စည်းထားတဲ့ dictionary ကနေ ပြုလုပ်ခြင်းဖြစ်တယ်။ 

```python
data = {"state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
        "year": [2000, 2001, 2002, 2001, 2002, 2003],
        "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
frame = pd.DataFrame(data)
```

```python
frame
```

ရလာတဲ့ DataFrame က index တွေဟာ Series တွေတုန်းကအတိုင်းပဲ အလိုအလျောက် သတ်မှတ်ပြီးသားဖြစ်မယ်။
column တွေကလည်း data(dictionary) မှာပါတဲ့ keys အစဉ်အတိုင်း သတ်မှတ်ပြီးသားဖြစ်မယ်။

note : Jupyter သုံးရင် pandas DataFrame object တွေကို ပို လှလှပပ ပြပေးတယ်။

အရွယ်အစား ကြီးမားတဲ့ DataFrame တွေအတွက်  `head` method ကို သုံးပြီး ပထမ ငါး row ပဲ ဆွဲထုတ်လို့ ရတယ်။

```python
frame.head()
```

အလားတူ `tail` ကို သုံးပြီး နောက်ဆုံး ငါး row ဆွဲထုတ်လို့ ရတယ်။

```python
frame.tail()
```

column တွေကို DataFrame ပြုလုပ်စဉ်မှာ သတ်မှတ်ပေးရင် သတ်မှတ်ပေးတဲ့အတိုင်း ရမယ်။

```python
pd.DataFrame(data, columns=["year", "state", "pop"])
```

dictionary ထဲမှာ မပါတဲ့ column ကို ထည့်ထားရင် အဲ့ column အတွက် တန်ဖိုးတွေအကုန် missing
value အနေနဲ့ ပြမယ်။

```python
frame2 = pd.DataFrame(data, columns=["year", "state", "pop", "debt"])
frame2
```

```python
frame2.index
```

```python
frame2.columns
```

DataFrame ထဲက column တွေကို Series မှာလိုပဲ dictionary-like
notation သို့မဟုတ် dot attribute notation ဖြင့် ရယူနိုင်တယ်။

```python
frame2["state"]
```

```python
frame2.year
```

Note:
- frame2[column] လို သုံးရင် ဘယ် column နဲ့ပဲ ဖြစ်ဖြစ် သုံးလို့ ရတယ်။ 
- ဒါပေမယ့် frame2.column မှာဆို column နာမည်တွေက DataFrame မှာပါတဲ့ အခြား method, attribute နာမည်တွေနဲ့ တူနေရင် မရဘူး။

ရလာတဲ့ column (Series) က DataFrame တစ်ခုလုံးရဲ့ index နဲ့ အတူတူပဲ၊ name
attribute လည်း သတ်မှတ်ပြီးသားဖြစ်တယ်။

```python
frame2
```

```python
# locate (by row name)
frame2.loc[1]
```

```python
# index locate (by position)
frame2.iloc[2]
```

Column တွေကို assignment(=) ဖြင့် ပြင်လို့ ရတယ်။ ဥပမာ လွတ်နေတဲ့ debt column ကို scalar value (ကိန်းဂဏန်း ၁ လုံး) သို့မဟုတ် array of values (ကိန်းအစုအဝေး) ဖြင့် ပြင်လို့ ရတယ်။

```python
frame2["debt"] = 16.5
frame2
```

```python
frame2["debt"] = np.arange(6.)
frame2
```

Series တစ်ခုခုနဲ့ assign လုပ်တဲ့ အခါမှာတော့ သူ့ရဲ့ index label က DataFrame ရဲ့ index နဲ့ ညှိသွားမှာ ဖြစ်ပြီး သက်ဆိုင်ရာ index မှာ တန်ဖိုးတွေ မရှိပါက missing value တွေနဲ့ ဖြည့်သွားမှာ ဖြစ်တယ်။

```python
val = pd.Series([-1.2, -1.5, -1.7], index=["two", "four", "five"])
frame2["debt"] = val
frame2
```

မရှိတဲ့ column တစ်ခုခုကို Assign ရင်တော့ အဲ့ column နာမည်နဲ့ column အသစ်တစ်ခု ပြုလုပ်မှာ ဖြစ်တယ်။

```python
frame2["eastern"] = frame2["state"] == "Ohio"
frame2
```

Note: frame2.eastern dot attribute notation ကို သုံးပြီးတော့ column အသစ် ပြုလုပ်လို့ မရပါ။

`del` keyword ကို သုံးပြီး dictionary မှာလို column တွေကို ဖျက်လို့ ရတယ်။ ဥပမာ - အပေါ်က "Ohio" ဖြစ်ရင် True မဟုတ်ရင် False လုပ်ထားတဲ့ Boolean တန်ဖိုးတွေပါတဲ့ eastern column ကို ဖျက်မယ်။

```python
del frame2["eastern"]
frame2.columns
```

Note:

DataFrame ကို indexing လုပ်ခြင်းဖြင့် ရလာတဲ့ column က view သဘောတရားဖြစ်ပြီး copy မဟုတ်ပါ။ ဆိုလိုတာက ရလာတဲ့ Series အပေါ်မှာ ပြုလုပ်လိုက်တဲ့ ပြောင်းလဲမှုတွေက မူလ DataFrame မှာပါ သက်ရောက်မှာ ဖြစ်တယ်။ သပ်သပ်စီ ခွဲထုတ်ချင်တာဆိုရင်တော့ Series ရဲ့ `copy` method ကို သုံးလို့ ရပါတယ်။

Table 5-1. Possible Data Inputs to the `DataFrame` Constructor

| **Type** | **Notes** |
|---------|-----------|
| 2D `ndarray` | A matrix of data, passing optional row and column labels |
| Dictionary of arrays, lists, or tuples | Each sequence becomes a column in the DataFrame; all sequences must be the same length |
| NumPy structured / record array | Treated as the *dictionary of arrays* case |
| Dictionary of `Series` | Each value becomes a column; indexes from each Series are unioned together to form the result’s row index if no explicit index is passed |
| Dictionary of dictionaries | Each inner dictionary becomes a column; keys are unioned to form the row index as in the *dictionary of Series* case |
| List of dictionaries or `Series` | Each item becomes a row in the DataFrame; unions of dictionary keys or Series indexes become the DataFrame’s column labels |
| List of lists or tuples | Treated as the *2D ndarray* case |
| Another `DataFrame` | The DataFrame’s indexes are used unless different ones are passed |
| NumPy `MaskedArray` | Like the *2D ndarray* case, except masked values are missing in the DataFrame result |aFrame result
